# 3. 데이터 조작하기 (DB: CASE WHEN)

```python
customerData["PRODUCTAGE_NEW"] = \
    np.where(customerData.PRODUCTAGE < 1, 1, customerData.PRODUCTAGE)

# 본인이 작성한 코드에 대한 검증 로직
# customerData.loc[ (customerData.PRODUCTAGE < 1) & (customerData.PRODUCTAGE_NEW != 1)]

customerData.loc[ (customerData.PRODUCTAGE >= 1) & (customerData.PRODUCTAGE_NEW != 1)]
```

```python
customerData["PRODUCTAGE_NEW"] = \
    np.where(customerData.PRODUCTAGE < 1, 1, 
             np.where(customerData.PRODUCTAGE < 2, 2,
                      np.where(customerData.PRODUCTAGE < 3, 3, customerData.PRODUCTAGE)))
```

```python
### 데이터 조작 pandas
### 사용하는 방법이 시점마다 다르기 때문에 ai는 람다 방식을 많이 사용한다 하지만 실제로는 현장에서는 잘 사용하지 않는다
### 1. np.where: 간단한 데이터 처리 시
### 2. loc: 딱 수정 포인트만 조건으로 검색해서 수정하고 빠지기
### 3. function: 복잡한 로직인 경우 함수를 정의해서 처리
### 4. dictionary 활용: 딕셔너리 정의 후 매핑: 문자 -> 숫자 변환
### 5. labelEncoder 활용: 카테고리 데이터를 숫자로 매핑( 근데 카테고리가 많은 경우)
### 6. onehotEncoder 활용: 인공지능에서 이미지 학습 1 -> 00001
```

```python
avgPriceMean = customerData.AVGPRICE.mean()
avgPriceMean
```

```python
%%timeit

### np.where 활용 방법: 데이터프레임["컬럼명"] = np.where(조건, 참일때 값, 거짓일때 값)
customerData["NEW_CAT"] = np.where(customerData.AVGPRICE < avgPriceMean, 0, 1)
customerData.loc[(customerData.NEW_CAT < 1) & (customerData.AVGPRICE >= avgPriceMean)]
```

```python
%%timeit

### loc 기능 (평균 가격보다 작게 구매 금액이 형성되어 있으면 -> 0)
customerData.loc[customerData.AVGPRICE < avgPriceMean, "NEW_CAT2"] = 0
customerData.loc[customerData.AVGPRICE >= avgPriceMean, "NEW_CAT2"] = 1
```

```python
customerData.loc[customerData.NEW_CAT != customerData.NEW_CAT2]
```

```python
### 3번째 로직을 특정 컬럼에 대해서 fastapi 호출도 해야하고 복잡하게 해야 할 때

def custCategory(inValue):
    ### avgPriceMean: 고객들의 평균 구매 가격
    # 각 고객들의 AVGPRICE 값
    # inValue = 4000
    # 리턴 Value
    outValue = 0
    
    # 각 고객들의 AVGPRICE가 전체 고객의 평균 AVGPRICE보다 작으면 0 아니면 1
    if inValue < avgPriceMean:
        outValue = 0
    else:
        outValue = 1
    return outValue
```

```python
customerData["NEW_CAT3"] = customerData.AVGPRICE.apply(custCategory)
```

```python
customerData.loc[(customerData.NEW_CAT2 != customerData.NEW_CAT3)]
```
