### 1. 타입 통합 및 특성 숫자 컬럼 추가

```python
# 1-1. 타입 통합 (스킵)
# 1-2. 특성 숫자 컬럼 추가
featuresData.HOLIDAY.drop_duplicates()
featuresData["HO_YH"] = np.where( featuresData.HOLIDAY=="Y", 1, 0)
featuresData["PRO_YH"] = np.where( featuresData.PROMOTION=="Y", 1, 0)
```

### 2. 특성 선정 및 데이터 분리

```python
corrDf = featuresData.corr(numeric_only=True)
stdCorr = 0.3
label = ["QTY"]
features = \
        list( corrDf.loc[ ( abs( corrDf.QTY) > stdCorr ) &  (corrDf.QTY != 1) ].index )
features

print(f"정답지는{label} 문제지는 { features}")
```

```python
# 머신러닝 라이브러리
from sklearn.model_selection import train_test_split
```

```python
# 2-2. 데이터 분리
trainData,testData = train_test_split( featuresData, test_size=0.2, random_state=5 )
print( featuresData.shape )
print( trainData.shape )
print( testData.shape )
trainDataFeatures = trainData.loc[ :, features ]
trainDataLabel = trainData.loc[ :, label]
testDataFeatures = testData.loc[ :, features ]
testDataLabel = testData.loc[:, label ]
```
