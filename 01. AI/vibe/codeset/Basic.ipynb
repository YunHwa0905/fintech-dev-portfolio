# 샘플 코드 작성
카멜케이스 변수명 및 지정된 반복문 스타일을 준수합니다.

```python
def processDataSample(inputList):
    """
    리스트 데이터를 처리하는 샘플 함수입니다.
    
    Args:
        inputList (list): 처리할 데이터 리스트
    """
    try:
        # 카멜케이스 변수명 사용
        resultList = []
        
        # 반복문: for i in range(0, len..) 형식 준수
        for i in range(0, len(inputList)):
            itemData = inputList[i]
            resultList.append(itemData)
            
        return resultList
    except Exception as e:
        return {"success": False, "message": str(e)}

# 데이터 실행 예시
sampleData = [1, 2, 3, 4, 5]
finalResult = processDataSample(sampleData)
print(finalResult)
```
